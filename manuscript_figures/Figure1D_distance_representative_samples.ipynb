```python
import os
import pandas as pd
import numpy as np
import scanpy as sc
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import importlib
import sys
sys.path.append("/Users/lingyi/Documents/spatial/colocalization/functions")

import tools
importlib.reload(tools)

```

```python
def identify_clusters_by_expression(adata, gene, z_threshold=0.5):
    """
    Identify clusters highly expressing a given gene using z-score thresholding.
    Returns clusters with expression > mean + z_threshold * std.
    """
    cluster_means = {}

    for cluster in adata.obs['clusters'].unique():
        mask = adata.obs['clusters'] == cluster
        
        if gene not in adata.var_names:
            cluster_means[cluster] = 0
            continue
        
        expr = adata[mask, gene].X
        if hasattr(expr, "toarray"):
            expr = expr.toarray()
        expr = np.array(expr).flatten()
        
        cluster_means[cluster] = np.mean(expr)

    # Convert to array for stats
    means = np.array(list(cluster_means.values()))
    mean_all = np.mean(means)
    std_all = np.std(means)

    # Choose clusters above threshold: mean + z_threshold * std; Pick clusters where expression is noticeably higher than normal. 
    high_clusters = [c for c, m in cluster_means.items() if m > (mean_all + z_threshold * std_all)]

    print(f"\n=== {gene} Expression Cluster Scan ===")
    print(f"Cluster Mean Expressions: {cluster_means}")
    print(f"Mean: {mean_all:.3f} | Std: {std_all:.3f} | Threshold: {mean_all + z_threshold*std_all:.3f}")
    print(f"High {gene} clusters: {high_clusters}")

    return high_clusters

def calculate_signature_score(adata, gene_list, score_name):
    """Calculate signature score as mean expression of genes in the list"""
    # Use scanpy's built-in subsetting to safely handle gene selection
    present_genes = [gene for gene in gene_list if gene in adata.var_names]
    
    if len(present_genes) == 0:
        raise ValueError(f"None of the genes in {gene_list} found in the dataset")
    
    # Use adata subsetting which handles duplicate gene names properly
    adata_subset = adata[:, present_genes]
    signature_expr = adata_subset.X
    
    if hasattr(signature_expr, 'toarray'):
        signature_expr = signature_expr.toarray()
    
    # Calculate mean across genes (axis=1 for mean across genes for each cell)
    signature_score = np.mean(signature_expr, axis=1)
    adata.obs[score_name] = signature_score
    
    return signature_score

def analyze_stromal_expression_by_tumor_proximity_annotation(adata, radius=200):
    """
    Compare aCAF expression in annotation-based stromal spots/cells near vs far from tumor spots/cells,
    and also analyze all *non-stromal, non-tumor* spots near vs far from tumors.
    """
    coords = adata.obsm['spatial']
    
    # Get tumor spot indices
    tumor_indices = np.where(adata.obs['annotation_tumor'])[0]
    tumor_coords = coords[tumor_indices]

    # For each spot, calculate distance to nearest tumor
    all_distances_to_tumor = []
    for i, coord in enumerate(coords):
        if len(tumor_coords) > 0:
            distances_to_tumors = np.sqrt(np.sum((tumor_coords - coord)**2, axis=1))
            min_distance = np.min(distances_to_tumors)
        else:
            min_distance = np.inf
        all_distances_to_tumor.append(min_distance)
    all_distances_to_tumor = np.array(all_distances_to_tumor)
    
    # Define masks
    stromal_mask = adata.obs['annotation_stromal'].values
    tumor_mask = adata.obs['annotation_tumor'].values
    other_cells_mask = ~(stromal_mask | tumor_mask)

    # Stromal cells near vs far from tumor
    stromal_near_tumor_mask = stromal_mask & (all_distances_to_tumor <= radius)
    stromal_far_tumor_mask = stromal_mask & (all_distances_to_tumor > radius)
    
    stromal_scores_near_tumor = adata.obs.loc[stromal_near_tumor_mask, 'aCAF_score'].values
    stromal_scores_far_tumor = adata.obs.loc[stromal_far_tumor_mask, 'aCAF_score'].values
    
    # compute for ALL OTHER (non-stromal, non-tumor) spots
    other_near_tumor_mask = other_cells_mask & (all_distances_to_tumor <= radius)
    other_far_tumor_mask = other_cells_mask & (all_distances_to_tumor > radius)

    other_scores_near_tumor = adata.obs.loc[other_near_tumor_mask, 'aCAF_score'].values
    other_scores_far_tumor = adata.obs.loc[other_far_tumor_mask, 'aCAF_score'].values
    
    # Get distances and stromal aCAF scores
    stromal_distances = all_distances_to_tumor[stromal_mask]
    stromal_acaf_scores = adata.obs.loc[stromal_mask, 'aCAF_score'].values
    
    return (stromal_scores_near_tumor, stromal_scores_far_tumor, 
            other_scores_near_tumor, other_scores_far_tumor, 
            stromal_distances, stromal_acaf_scores)

def cohens_d(group1, group2):
    """Calculate Cohen's d effect size"""
    if len(group1) == 0 or len(group2) == 0:
        return np.nan
    pooled_std = np.sqrt(((len(group1) - 1) * np.var(group1) + (len(group2) - 1) * np.var(group2)) / 
                        (len(group1) + len(group2) - 2))
    if pooled_std == 0:
        return np.nan
    return (np.mean(group1) - np.mean(group2)) / pooled_std

def process_sample(dataset_name, sampleID, base_folder="/Volumes/T7/datasets"):
    """Process a single sample and return analysis results"""
    print(f"\n===== {dataset_name} | {sampleID} =====")
    
    # Construct folder path
    folder = f"{base_folder}/{dataset_name}/raw/{sampleID}"
    
    try:
        # Load and process data
        proc = tools.spatialTool()
        adata = proc.dataprocess(folder, skip_qc=False, verbose = False)
        
        # Automatically identify fibroblast clusters (high LUM expression)
        fibroblast_clusters = identify_clusters_by_expression(adata, 'LUM', z_threshold=0.5)
        
        # Automatically identify epithelial clusters (high EPCAM expression)  
        epithelial_clusters = identify_clusters_by_expression(adata, 'EPCAM', z_threshold=0.5)
        
        # Assign annotations based on identified clusters
        epithelial_spots = adata.obs['clusters'].isin(epithelial_clusters)
        stromal_spots = adata.obs['clusters'].isin(fibroblast_clusters)
        
        adata.obs['annotation_tumor'] = epithelial_spots
        adata.obs['annotation_stromal'] = stromal_spots
        
        print(f"Epithelial clusters: {epithelial_clusters}")
        print(f"Fibroblast clusters: {fibroblast_clusters}")
        print(f"Epithelial spots: {epithelial_spots.sum()}")
        print(f"Stromal spots: {stromal_spots.sum()}")
        
        # Calculate aCAF signature score
        acaf_genes = ['COL11A1', 'THBS2', 'INHBA', 'POSTN', 'SPARC', 'MMP11', 'COL10A1']
        # acaf_genes = ['COL11A1']
        present_acaf_genes = [gene for gene in acaf_genes if gene in adata.var_names]
        print(f"Present aCAF genes: {present_acaf_genes}")
        
        if len(present_acaf_genes) > 0:
            acaf_score = calculate_signature_score(adata, present_acaf_genes, 'aCAF_score')
        else:
            print("No aCAF genes found in dataset, skipping...")
            return None, None
        
        # Perform distance analysis
        set_radius = 200
        (stromal_near, stromal_far, other_near, other_far, 
         stromal_distances, stromal_acaf_scores) = analyze_stromal_expression_by_tumor_proximity_annotation(adata, radius=set_radius)
        
        # Calculate statistics if we have stromal data
        if len(stromal_distances) > 0:
            correlation_stromal, p_corr_stromal = stats.pearsonr(stromal_distances, stromal_acaf_scores)
            spearman_corr_stromal, p_spearman_stromal = stats.spearmanr(stromal_distances, stromal_acaf_scores)
            
            print(f"Stromal correlation analysis:")
            print(f"- Pearson correlation: r={correlation_stromal:.3f}, p={p_corr_stromal:.6f}")
            print(f"- Spearman correlation: ρ={spearman_corr_stromal:.3f}, p={p_spearman_stromal:.6f}")
            
            # Statistical test for near vs far
            if len(stromal_near) > 0 and len(stromal_far) > 0:
                stat1, p_val1 = stats.mannwhitneyu(stromal_near, stromal_far, alternative='two-sided')
                effect_size1 = cohens_d(stromal_near, stromal_far)
                print(f"Mann-Whitney U test: U={stat1:.2f}, p={p_val1:.6f}")
                print(f"Effect size (Cohen's d): {effect_size1:.3f}")
        else:
            correlation_stromal, p_corr_stromal = np.nan, np.nan
            spearman_corr_stromal, p_spearman_stromal = np.nan, np.nan
            print("No stromal spots found for analysis")
        
        return adata, {
            'dataset': dataset_name,
            'sample': sampleID,
            'epithelial_clusters': epithelial_clusters,
            'fibroblast_clusters': fibroblast_clusters,
            'n_epithelial': epithelial_spots.sum(),
            'n_stromal': stromal_spots.sum(),
            'stromal_distances': stromal_distances,
            'stromal_acaf_scores': stromal_acaf_scores,
            'pearson_r': correlation_stromal,
            'pearson_p': p_corr_stromal,
            'spearman_r': spearman_corr_stromal,
            'spearman_p': p_spearman_stromal
        }
        
    except Exception as e:
        print(f"Error processing {dataset_name}/{sampleID}: {str(e)}")
        return None, None

def create_distance_plot(results_dict, dataset_name, sampleID):
    """Create distance vs aCAF plot for a single sample with robust error handling"""
    if results_dict is None:
        return None
        
    stromal_distances = results_dict['stromal_distances']
    stromal_acaf_scores = results_dict['stromal_acaf_scores']
    
    if len(stromal_distances) == 0:
        return None
    
    fig, ax = plt.subplots(1, 1, figsize=(8, 6))
    
    # Scatter plot
    ax.scatter(stromal_distances, stromal_acaf_scores, alpha=0.7, s=10, color='red')
    
    # Add trend line with robust error handling
    if len(stromal_distances) > 1:
        try:
            # Check for data variability before fitting
            if np.var(stromal_distances) > 1e-10 and np.var(stromal_acaf_scores) > 1e-10:
                z = np.polyfit(stromal_distances, stromal_acaf_scores, 1)
                p = np.poly1d(z)
                x_trend = np.linspace(stromal_distances.min(), stromal_distances.max(), 100)
                ax.plot(x_trend, p(x_trend), "blue", alpha=0.8, linewidth=2)
            else:
                print(f"Warning: Insufficient data variability for trend line in {sampleID}")
        except np.linalg.LinAlgError as e:
            print(f"Warning: Could not fit trend line for {sampleID}: {e}")
        except Exception as e:
            print(f"Warning: Unexpected error fitting trend line for {sampleID}: {e}")
    
    ax.set_xlabel('Distance to nearest tumor')
    ax.set_ylabel('aCAF score')
    
    # Safe formatting of correlation values
    try:
        pearson_r = results_dict["pearson_r"]
        pearson_p = results_dict["pearson_p"]
        if not np.isnan(pearson_r) and not np.isnan(pearson_p):
            title = f'{sampleID}\nStromal aCAF score vs Distance to Nearest Tumor\nr={pearson_r:.3f}, p={pearson_p:.3e}'
        else:
            title = f'{sampleID}\nStromal aCAF score vs Distance to Nearest Tumor\n(correlation unavailable)'
    except:
        title = f'{sampleID}\nStromal aCAF score vs Distance to Nearest Tumor'
        
    ax.set_title(title, fontweight='bold')
    ax.grid(True, alpha=0.3)
    
    return fig

# Main processing loop
all_results = []
failed_samples = []

print(f"Processing {len(df_filter_high_aCAF)} samples...")

for idx, row in df_filter_high_aCAF.iterrows():
    dataset_name = row['dataset_name']
    sampleID = row['SampleID']
    
    adata, results = process_sample(dataset_name, sampleID)
    
    if results is not None:
        all_results.append(results)
        
        # Create and show plot for this sample
        fig = create_distance_plot(results, dataset_name, sampleID)
        if fig is not None:
            plt.show()
        else:
            print(f"No plot generated for {sampleID} (no stromal data)")
    else:
        failed_samples.append(f"{dataset_name}/{sampleID}")

print(f"\nProcessing complete!")
print(f"Successfully processed: {len(all_results)} samples")
print(f"Failed samples: {len(failed_samples)}")
if failed_samples:
    print("Failed:", failed_samples)
```
