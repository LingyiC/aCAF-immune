# Figure 2C


Plots of the mean levels of the marker expression and the distance to the tumor region.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
def df_bin(X, Y, total_bins = 10):
  # 2. Define the bin ranges (edges)
  bins = np.linspace(X.min(), X.max(), total_bins + 1)
  # 3. Digitize the data to assign each point to a bin
  idx = np.digitize(X, bins)
  # 4. Calculate the median of Y values within each bin
  running_median = [np.mean(Y[idx == k]) for k in range(1, total_bins + 1)]
  # 5. Calculate the bin centers for the x-axis plot points
  delta = bins[1] - bins[0]
  bin_centers = bins[:-1] + delta/2

  return bin_centers, running_median
```

```python
path = "~/nontumor_distance_expression_export_csv/"
```

```python
file = ["CID4465_spatial","SN124_A938797_Rep2","HT232P1H2A2","GSM7498813_SS1945070_processed"]
```

```python
fig = plt.figure(figsize=(12, 3))
nrows, ncols = 1, 5

for idx in range(len(file)):

  csv_path= path + file[idx] + "_nontumor_data.csv"

  df = pd.read_csv(csv_path).transpose()
  df.columns = df.iloc[0,:]
  df_cleaned = df.drop('Unnamed: 0')

  x_COL11A1, y_COL11A1= df_bin(df_cleaned['distance_to_tumor'], df_cleaned['COL11A1'])
  x_CD3D, y_CD3D= df_bin(df_cleaned['distance_to_tumor'], df_cleaned['CD3D'])
  x_CD8A, y_CD8A= df_bin(df_cleaned['distance_to_tumor'], df_cleaned['CD8A'])
  x_FOXP3, y_FOXP3= df_bin(df_cleaned['distance_to_tumor'], df_cleaned['FOXP3'])
  x_CD163, y_CD163= df_bin(df_cleaned['distance_to_tumor'], df_cleaned['CD163'])

  ax = fig.add_subplot(nrows, ncols, idx+1)

  ax.plot(x_COL11A1, y_COL11A1, 'k-', lw=3, marker='', markersize=1, label='COL11A1 aCAF')
  ax.plot(x_CD163, y_CD163, 'b-', lw=3, marker='', markersize=1, label='CD163 Macrophage')
  ax.plot(x_CD3D, y_CD3D, 'r-', lw=3, marker='', markersize=1, label='CD3D T')
  ax.plot(x_CD8A, y_CD8A, 'y-', lw=3, marker='', markersize=1, label='CD8A Cytotoxic T')
  ax.plot(x_FOXP3, y_FOXP3, 'g-', lw=3, marker='', markersize=1, label='FOXP3 Treg')
  ax.set_title(file[idx])
  if(idx==0):
    ax.set_xlabel('Distance to Tumor (µm)')
    ax.set_ylabel('Mean Expression')
    #ax.legend()

  ax.axis('tight')
  ax.grid(True)

plt.tight_layout()

plt.show()
```
